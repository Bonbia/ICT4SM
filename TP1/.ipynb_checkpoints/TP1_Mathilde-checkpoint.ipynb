```python
Init_Time = res = pb_coll.find({},{'init_time':1,'_id':0}).sort("init_time", 1).limit(1)
Final_Time = pb_coll.find({},{'final_time':1,'_id':0}).sort("init_time", -1).limit(1)

Init_Time = next(Init_Time)['init_time']
Final_Time = next(Final_Time)['final_time']

print(Init_Time, Final_Time)
```

```python
pipeline = [
    {
        "$match": {
            "city": "Torino",
            "init_time": {
                "$gte": Init_Time,
                "$lte": Final_Time
            }
        }
    },
    {
        "$group": {
            "_id": {
                "week": {
                    "$dateTrunc": {
                        "date": {
                            "$toDate": {
                                "$multiply": ["$init_time", 1000]
                            }
                        },
                        "unit": "week",
                        "startOfWeek": "monday"
                    }
                },
                "plate": "$plate"
            }
        }
    },
    {
        "$group": {
            "_id": "$_id.week",
            "nb_plates": {"$sum": 1}
        }
    },
    {
        "$sort": {
            "_id": 1
        }
    }
]

weekly = list(pb_coll.aggregate(pipeline))
```

```python
df = pd.DataFrame(weekly)

df.rename(columns={
    "_id": "week",
    "nb_plates": "unique_plates"
}, inplace=True)

df
```

## Q2: Consider the system utilization over time: 
aggregate rentals per hour of the day, and then plot the number of booked/parked cars (or percentage of booked/parked cars) per hour versus time of day.
Do you notice any outliers? Can you explain them?

V1 : We count a rental on all the hour time slot it appears 
V2 : We count a rental only on the hour it begins in 

```python
def avg_systutilization(coll, city):

    start_ts = datetime.datetime(2017, 11, 1, 0, 0, 0)
    end_ts = datetime.datetime(2018, 2, 1, 0, 0, 0)

    df = pd.DataFrame(list(coll.find(
        {'city': city,
         "init_date": {
             "$gte": start_ts,
             "$lt": end_ts
         }},
        {
            '_id': 0,
            'init_date': 1,
            'final_date': 1
        }
    )))

    if df.empty:
        print("No data available for this city and period.")
        return

    # Générer les heures de la série temporelle
    hours = pd.date_range(
        df['init_date'].min().floor('h'),
        df['final_date'].max().floor('h'),
        freq='h'
    )

    # V1 : nombre de locations actives à chaque heure
    hourly_rentals_v1 = pd.Series({
        hour: (
            (df['init_date'] < hour + pd.Timedelta(hours=1)) &
            (df['final_date'] > hour)
        ).sum()
        for hour in hours
    })

    # Regrouper par heure de la journée (0 à 23)
    hourly_profile_v1 = hourly_rentals_v1.groupby(
        hourly_rentals_v1.index.hour
    ).mean().reindex(range(24), fill_value=0)

    # Tracer le graphique
    hourly_profile_v1.plot(
        kind='bar',
        figsize=(12, 5),
        rot=0
    )

    plt.xlabel('Hour of day')
    plt.ylabel('Average number of active rentals')
    plt.title(f'Average system utilization by hour — {city}')
    plt.xticks(range(24))
    plt.grid(axis='y')
    plt.show()

    return hourly_profile_v1
```

```python
def avg_parking_start(coll,city):

    start_ts = datetime.datetime(2017, 11, 1, 0, 0, 0)
    end_ts = datetime.datetime(2018, 2, 1, 0, 0, 0)

    df = pd.DataFrame(list(coll.find(
        {'city': city,
         "init_date": {
             "$gte": start_ts,
             "$lt": end_ts
         }},
        {
            '_id': 0,
            'init_date': 1
        }
    )))

    # Extraire le jour et l'heure de chaque location
    df['day'] = df['init_date'].dt.date
    df['hour'] = df['init_date'].dt.hour
    
    # Compter les locations pour chaque jour et chaque heure
    rentals_per_day_hour = (
        df.groupby(['day', 'hour'])
          .size()
          .rename('rentals')
          .reset_index()
    )
    
    # Calculer la moyenne par heure sur l'ensemble des jours
    hourly_rentals_v2 = (
        rentals_per_day_hour.groupby('hour')['rentals']
        .mean()
        .reindex(range(24), fill_value=0)
    )
    
    # Tracer le graphique
    hourly_rentals_v2.plot(
        kind='bar',
        figsize=(12, 5),
        rot=0
    )
    
    plt.xlabel('Hour of day')
    plt.ylabel('Average number of rentals started')
    plt.title(f'Average rental starts by hour of day — {city}')
    plt.xticks(range(24))
    plt.grid(axis='y')
    plt.show()
```

```python
def fleet_utilization_city(coll,city):

    if city == 'Columbus':
        start, end = start_Col, end_Col
    else:
        start, end = start_ToFr, end_ToFr

    start_ts = int(start.timestamp())
    end_ts = int(end.timestamp())

    df = pd.DataFrame(list(coll.find(
        {'city': city,
        "init_time": {
                    "$gte": start_ts,
                    "$lte": end_ts
                }},
        {
            '_id': 0,
            'init_date': 1,
            'final_date': 1,
            'plate': 1
        }
    )))
    
    # Convertir les dates
    df['init_date'] = pd.to_datetime(df['init_date'])
    df['final_date'] = pd.to_datetime(df['final_date'])
    
    # Garder les locations valides
    df = df.dropna(subset=['init_date', 'final_date'])
    df = df[df['final_date'] > df['init_date']].copy()
    
    if df.empty:
        raise ValueError("Aucune location valide trouvée pour Torino.")
    
    # Nombre de voitures distinctes observées dans pb_coll
    total_cars = df.loc[
        df['plate'].notna() & (df['plate'] != -1),
        'plate'
    ].nunique()
    
    if total_cars == 0:
        raise ValueError("Aucune plaque valide trouvée dans les données.")
    
    # Créer les créneaux horaires
    hours = pd.date_range(
        df['init_date'].min().floor('h'),
        df['final_date'].max().ceil('h'),
        freq='h',
        inclusive='left'
    )
    
    # V1 : une location compte dans chaque heure qu'elle chevauche
    hourly_rentals = pd.Series({
        hour: (
            (df['init_date'] < hour + pd.Timedelta(hours=1)) &
            (df['final_date'] > hour)
        ).sum()
        for hour in hours
    })
    
    # Moyenne par heure de la journée sur tous les jours observés
    hourly_avg = (
        hourly_rentals.groupby(hourly_rentals.index.hour)
        .mean()
        .reindex(range(24), fill_value=0)
    )
    
    # Pourcentage approximatif de voitures occupées
    utilization_pct = hourly_avg / total_cars * 100
    
    print("Nombre de locations valides :", len(df))
    print("Nombre de voitures distinctes observées :", total_cars)
    print("Valeurs manquantes :", utilization_pct.isna().sum())
    print("Valeurs infinies :", np.isinf(utilization_pct).sum())
    print(utilization_pct.round(2))
    
    # Tracer
    utilization_pct.plot(
        kind='bar',
        figsize=(12, 5),
        rot=0
    )
    
    plt.xlabel('Hour of day')
    plt.ylabel('Booked cars (%)')
    plt.title(f'Average fleet utilization by hour — {city}')
    plt.xticks(range(24))
    plt.grid(axis='y')
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

def daily_duration_stats(data):
    df = pd.concat(data.values(), ignore_index=True).copy()

    # Convertir les dates et les durées
    df["init_time"] = pd.to_datetime(df["init_time"], errors="coerce")
    df["duration_min"] = pd.to_numeric(df["duration_min"], errors="coerce")

    df = df.dropna(subset=["init_time", "duration_min"])
    df = df[df["duration_min"] >= 0]

    # Regrouper par jour
    df["day"] = df["init_time"].dt.date

    # Calculer les statistiques
    stats = df.groupby("day")["duration_min"].agg(
        count="count",
        mean="mean",
        median="median",
        std="std"
    )

    # Calculer les percentiles
    percentiles = df.groupby("day")["duration_min"].quantile(
        [0.25, 0.50, 0.75, 0.90, 0.95]
    ).unstack()

    percentiles.columns = ["p25", "p50", "p75", "p90", "p95"]

    return stats.join(percentiles)


booking_stats = daily_duration_stats(bookings_f)
parking_stats = daily_duration_stats(parkings_f)

display(booking_stats.head())
display(parking_stats.head())
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def plot_daily_duration_stats(
    bookings,
    parkings,
    cities=None,
    duration_col="duration_min",
    percentiles=(10, 25, 75, 90),
):
    """
    Plot daily duration statistics for bookings and parkings.

    Parameters
    ----------
    bookings, parkings : dict[str, DataFrame]
        Dictionnaires {ville: DataFrame}.
    cities : list[str] or None
        Villes à afficher. Par défaut, toutes les villes communes.
    duration_col : str
        Colonne contenant la durée en minutes.
    percentiles : tuple
        Percentiles à calculer.

    Returns
    -------
    stats : dict
        Statistiques journalières par type et par ville.
    """
    if cities is None:
        cities = sorted(set(bookings) & set(parkings))

    stats = {"bookings": {}, "parkings": {}}

    for kind, datasets in [
        ("bookings", bookings),
        ("parkings", parkings),
    ]:
        for city in cities:
            df = datasets[city].copy()

            if duration_col not in df.columns:
                raise ValueError(
                    f"Colonne '{duration_col}' absente pour {kind}, {city}"
                )

            # Reconstruire la date si nécessaire
            if "day" not in df.columns:
                if "init_time" not in df.columns:
                    raise ValueError(
                        f"Il faut une colonne 'day' ou 'init_time' "
                        f"pour {kind}, {city}"
                    )
                df["day"] = pd.to_datetime(
                    df["init_time"], unit="s", utc=True
                ).dt.strftime("%Y-%m-%d")

            df[duration_col] = pd.to_numeric(
                df[duration_col], errors="coerce"
            )
            df = df[
                df[duration_col].notna()
                & np.isfinite(df[duration_col])
                & (df[duration_col] > 0)
                & df["day"].notna()
            ]

            # Statistiques par jour
            grouped = df.groupby("day")[duration_col]

            daily = grouped.agg(
                count="count",
                average="mean",
                median="median",
                standard_deviation="std",
            )

            for p in percentiles:
                daily[f"p{p}"] = grouped.quantile(p / 100)

            daily.index = pd.to_datetime(daily.index)
            daily = daily.sort_index()

            stats[kind][city] = daily

            # Graphiques
            fig, axes = plt.subplots(
                3, 1, figsize=(14, 11), sharex=True
            )

            fig.suptitle(
                f"{kind.capitalize()} — {city}: daily duration statistics",
                fontsize=14,
            )

            # 1. Moyenne et médiane
            axes[0].plot(
                daily.index, daily["average"],
                label="Average", linewidth=1.5
            )
            axes[0].plot(
                daily.index, daily["median"],
                label="Median", linewidth=1.5
            )
            axes[0].set_ylabel("Duration (min)")
            axes[0].set_title("Average and median")
            axes[0].legend()
            axes[0].grid(True, alpha=0.3)

            # 2. Écart-type
            axes[1].plot(
                daily.index, daily["standard_deviation"],
                label="Standard deviation",
            )
            axes[1].set_ylabel("Duration (min)")
            axes[1].set_title("Standard deviation")
            axes[1].legend()
            axes[1].grid(True, alpha=0.3)

            # 3. Percentiles
            for p in percentiles:
                axes[2].plot(
                    daily.index, daily[f"p{p}"],
                    label=f"P{p}",
                )

            axes[2].set_ylabel("Duration (min)")
            axes[2].set_xlabel("Day")
            axes[2].set_title("Duration percentiles")
            axes[2].legend(ncol=len(percentiles))
            axes[2].grid(True, alpha=0.3)

            plt.tight_layout()
            plt.show()

    return stats
```

```python
daily_stats = plot_daily_duration_stats(
    bookings=bookings_f,
    parkings=parkings_f,
    cities=["Torino", "Columbus", "Frankfurt"],
)
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates


def plot_daily_duration_stats_compact(
    bookings,
    parkings,
    cities=None,
    duration_col="duration_min",
    percentiles=(10, 25, 75, 90),
):
    """
    Calcule les statistiques journalières et produit des graphiques compacts
    pour les bookings et parkings de plusieurs villes.
    """
    if cities is None:
        cities = sorted(set(bookings) & set(parkings))
    else:
        cities = list(cities)

    stats = {"bookings": {}, "parkings": {}}

    for kind, datasets in [
        ("bookings", bookings),
        ("parkings", parkings),
    ]:
        for city in cities:
            if city not in datasets:
                print(f"Ville {city} absente de {kind}, ignorée.")
                continue

            df = datasets[city].copy()

            if duration_col not in df.columns:
                raise ValueError(
                    f"Colonne '{duration_col}' absente pour {kind}, {city}"
                )

            if "day" not in df.columns:
                if "init_time" not in df.columns:
                    raise ValueError(
                        f"Il faut une colonne 'day' ou 'init_time' "
                        f"pour {kind}, {city}"
                    )
                df["day"] = pd.to_datetime(
                    df["init_time"], unit="s", utc=True
                ).dt.strftime("%Y-%m-%d")

            df[duration_col] = pd.to_numeric(
                df[duration_col], errors="coerce"
            )

            df = df[
                df[duration_col].notna()
                & np.isfinite(df[duration_col])
                & (df[duration_col] > 0)
                & df["day"].notna()
            ]

            grouped = df.groupby("day")[duration_col]

            daily = grouped.agg(
                count="count",
                average="mean",
                median="median",
                standard_deviation="std",
            )

            for p in percentiles:
                daily[f"p{p}"] = grouped.quantile(p / 100)

            daily.index = pd.to_datetime(daily.index)
            daily = daily.sort_index()

            stats[kind][city] = daily

            # Graphique compact : deux panneaux
            fig, axes = plt.subplots(
                2, 1,
                figsize=(11, 6),
                sharex=True,
                gridspec_kw={"height_ratios": [1, 1.2]},
            )

            fig.suptitle(
                f"{kind.capitalize()} — {city}",
                fontsize=13,
                fontweight="bold",
            )

            # 1. Moyenne et médiane
            axes[0].plot(
                daily.index, daily["average"],
                label="Moyenne",
                color="tab:blue",
                linewidth=1.5,
            )
            axes[0].plot(
                daily.index, daily["median"],
                label="Médiane",
                color="tab:orange",
                linewidth=1.5,
            )

            axes[0].set_ylabel("Durée (min)")
            axes[0].legend(frameon=False, ncol=2)
            axes[0].grid(True, alpha=0.25)

            # 2. Médiane et intervalle interquartile
            axes[1].plot(
                daily.index, daily["median"],
                label="Médiane",
                color="tab:orange",
                linewidth=1.5,
            )

            if 25 in percentiles and 75 in percentiles:
                axes[1].fill_between(
                    daily.index,
                    daily["p25"].to_numpy(dtype=float),
                    daily["p75"].to_numpy(dtype=float),
                    color="tab:orange",
                    alpha=0.2,
                    label="P25–P75",
                )

            axes[1].set_ylabel("Durée (min)")
            axes[1].set_xlabel("Date")
            axes[1].legend(frameon=False, ncol=2)
            axes[1].grid(True, alpha=0.25)

            for ax in axes:
                ax.spines["top"].set_visible(False)
                ax.spines["right"].set_visible(False)

            axes[1].xaxis.set_major_locator(
                mdates.AutoDateLocator(minticks=4, maxticks=8)
            )
            axes[1].xaxis.set_major_formatter(
                mdates.ConciseDateFormatter(
                    axes[1].xaxis.get_major_locator()
                )
            )

            fig.tight_layout()
            plt.show()

    return stats
```

```python
daily_stats = plot_daily_duration_stats_compact(
    bookings=bookings_f,
    parkings=parkings_f,
    cities=["Torino", "Columbus", "Frankfurt"],
)
```

Several trends and changes can be observed over the period from November 2017 to January 2018.

```python
!pip install holidays
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import holidays


def plot_daily_duration_stats_holiweekends(
    bookings,
    parkings,
    cities=None,
    duration_col="duration_min",
    percentiles=(10, 25, 75, 90),
    rolling_window=7,
    show_weekends=True,
    show_holidays=True,
    holiday_dates=None,
):
    """
    Daily duration analysis:
    - Average, median and rolling trends
    - Standard deviation and percentiles
    - Weekend and public-holiday annotations
    - Comparison of weekdays, weekends and holidays
    - Detection of unusually long durations

    holiday_dates:
        Optional dict {city: {date: holiday_name}}.
        Dates should use YYYY-MM-DD format.
        Custom dates are added to the automatic calendar.
    """

    if cities is None:
        cities = sorted(set(bookings) & set(parkings))

    # Country/subdivision mapping for the cities
    city_calendars = {
        "turin": ("IT", "TO"),
        "torino": ("IT", "TO"),
        "columbus": ("US", "OH"),
        "frankfurt": ("DE", "HE"),
    }

    holiday_dates = holiday_dates or {}
    stats = {
        "bookings": {},
        "parkings": {},
        "comparison": {"bookings": {}, "parkings": {}},
    }

    for kind, datasets in [
        ("bookings", bookings),
        ("parkings", parkings),
    ]:
        for city in cities:
            df = datasets[city].copy()

            if duration_col not in df.columns:
                raise ValueError(
                    f"Missing '{duration_col}' for {kind}, {city}"
                )

            # Create daily date if necessary
            if "day" not in df.columns:
                if "init_time" not in df.columns:
                    raise ValueError(
                        f"Missing 'day' or 'init_time' for {kind}, {city}"
                    )

                df["day"] = pd.to_datetime(
                    df["init_time"], unit="s", utc=True
                ).dt.strftime("%Y-%m-%d")

            df["day"] = pd.to_datetime(df["day"], errors="coerce").dt.normalize()
            df[duration_col] = pd.to_numeric(
                df[duration_col], errors="coerce"
            )

            df = df[
                df["day"].notna()
                & df[duration_col].notna()
                & np.isfinite(df[duration_col])
                & (df[duration_col] > 0)
            ].copy()

            if df.empty:
                print(f"No valid data for {kind}, {city}")
                continue

            # Daily statistics
            grouped = df.groupby("day")[duration_col]

            daily = grouped.agg(
                count="count",
                average="mean",
                median="median",
                standard_deviation="std",
            )

            for p in percentiles:
                daily[f"p{p}"] = grouped.quantile(p / 100)

            daily = daily.sort_index()

            # Rolling average helps reveal longer-term trends
            daily["average_rolling"] = daily["average"].rolling(
                rolling_window, min_periods=1
            ).mean()

            # Generate the appropriate public holiday calendar
            first_year = int(daily.index.min().year)
            last_year = int(daily.index.max().year)
            city_key = city.strip().lower()

            calendar_info = city_calendars.get(city_key)
            if calendar_info:
                country, subdivision = calendar_info
                calendar = holidays.country_holidays(
                    country,
                    subdiv=subdivision,
                    years=range(first_year, last_year + 1),
                )
                holiday_map = {
                    pd.Timestamp(date): name
                    for date, name in calendar.items()
                }
            else:
                holiday_map = {}

            # Add user-defined holidays or local dates
            custom = holiday_dates.get(city, {})
            for date, name in custom.items():
                holiday_map[pd.Timestamp(date).normalize()] = name

            daily["is_weekend"] = daily.index.dayofweek >= 5
            daily["holiday"] = [
                holiday_map.get(date, None) for date in daily.index
            ]
            daily["is_holiday"] = daily["holiday"].notna()

            daily["day_type"] = np.select(
                [
                    daily["is_holiday"],
                    daily["is_weekend"],
                ],
                [
                    "Public holiday",
                    "Weekend",
                ],
                default="Weekday",
            )

            stats[kind][city] = daily

            # Compare daily statistics across day types
            comparison = (
                daily.groupby("day_type")[
                    ["average", "median", "standard_deviation", "count"]
                ]
                .agg(["mean", "median", "count"])
            )

            stats["comparison"][kind][city] = comparison

            print(f"\n{'=' * 60}")
            print(f"{kind.upper()} — {city}")
            print("Daily statistics by day type:")
            print(
                daily.groupby("day_type")[
                    ["average", "median", "standard_deviation"]
                ].mean().round(2)
            )

    return stats
```

```python
daily_stats = plot_daily_duration_stats_holiweekends(
    bookings=bookings_f,
    parkings=parkings_f,
    cities=["Torino", "Columbus", "Frankfurt"],
)
```
